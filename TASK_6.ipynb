Create Sample Dataset

````python
import pandas as pd

data = {
    "Employee_ID": [101,102,103,104,105,106,107,108,109,110],
    "Name": [
        "Rahul","Priya","Amit","Sneha","Karan",
        "Pooja","Riya","Arjun","Neha","Vikram"
    ],
    "Department": [
        "IT","HR","Finance","IT","Sales",
        "HR","Finance","IT","Sales","IT"
    ],
    "Age": [25,28,30,27,35,29,24,31,26,32],
    "Salary": [50000,45000,60000,55000,48000,47000,52000,70000,49000,65000],
    "Experience": [2,3,6,4,8,5,2,9,3,7],
    "City": [
        "Bhubaneswar","Cuttack","Puri","Bhubaneswar",
        "Rourkela","Cuttack","Balasore","Puri",
        "Sambalpur","Bhubaneswar"
    ]
}

df = pd.DataFrame(data)

df.to_csv("employee_data.csv", index=False)

print(df)
````

Load Dataset

````python
df = pd.read_csv("employee_data.csv")

df.head()
````

Display Dataset Information

````python
print(df.info())

print("\nShape of Dataset:", df.shape)
````

Select Specific Columns

````python
selected_columns = df[["Name", "Department", "Salary"]]

print(selected_columns)
````

Filter Rows (Salary Greater Than 50,000)

````python
high_salary = df[df["Salary"] > 50000]

print(high_salary)
````

Filter Employees from IT Department

````python
it_department = df[df["Department"] == "IT"]

print(it_department)
````

Multiple Conditions

````python
filtered_data = df[
    (df["Department"] == "IT") &
    (df["Salary"] > 55000)
]

print(filtered_data)
````

Employees with Experience Greater Than 5 Years

````python
experienced = df[df["Experience"] > 5]

print(experienced)
````

Sort Dataset by Salary (Ascending)

````python
salary_ascending = df.sort_values(by="Salary")

print(salary_ascending)
````

Sort Dataset by Salary (Descending)

````python
salary_descending = df.sort_values(
    by="Salary",
    ascending=False
)

print(salary_descending)
````

Sort by Department and Salary

````python
sorted_data = df.sort_values(
    by=["Department", "Salary"],
    ascending=[True, False]
)

print(sorted_data)
````

Create Final Filtered Dataset

````python
final_dataset = df[
    (df["Department"] == "IT") &
    (df["Salary"] >= 55000)
][["Employee_ID", "Name", "Department", "Salary"]]

print(final_dataset)
````

Save Filtered Dataset

````python
final_dataset.to_csv(
    "filtered_employee_data.csv",
    index=False
)

print("Filtered dataset saved successfully.")
````
