Create Sample Business Dataset

```python
import pandas as pd

data = {
    "Order_ID":[1001,1002,1003,1004,1005,1006,1007,1008,1009,1010],
    "Product":[
        "Laptop",
        "Mouse",
        "Keyboard",
        "Monitor",
        "Laptop",
        "Headphones",
        "Mouse",
        "Laptop",
        "Keyboard",
        "Monitor"
    ],
    "Category":[
        "Electronics",
        "Accessories",
        "Accessories",
        "Electronics",
        "Electronics",
        "Accessories",
        "Accessories",
        "Electronics",
        "Accessories",
        "Electronics"
    ],
    "City":[
        "Bhubaneswar",
        "Cuttack",
        "Puri",
        "Rourkela",
        "Bhubaneswar",
        "Cuttack",
        "Balasore",
        "Puri",
        "Sambalpur",
        "Bhubaneswar"
    ],
    "Quantity":[2,5,3,1,4,6,2,1,5,2],
    "Price":[55000,700,1200,15000,56000,2500,750,54000,1300,15500],
    "Sales":[110000,3500,3600,15000,224000,15000,1500,54000,6500,31000]
}

df = pd.DataFrame(data)

df.to_csv("sales_data.csv", index=False)

print(df)
```

Load Dataset

```python
df = pd.read_csv("sales_data.csv")

df.head()
```

Business Analysis

```python
total_sales = df["Sales"].sum()
average_sales = df["Sales"].mean()
maximum_sale = df["Sales"].max()
minimum_sale = df["Sales"].min()

top_product = df.groupby("Product")["Sales"].sum().idxmax()

top_city = df.groupby("City")["Sales"].sum().idxmax()

top_category = df.groupby("Category")["Sales"].sum().idxmax()

total_orders = df["Order_ID"].count()
```

Display Summary

```python
print("===== BUSINESS SUMMARY =====")

print("Total Sales :", total_sales)

print("Average Sales :", round(average_sales,2))

print("Highest Sale :", maximum_sale)

print("Lowest Sale :", minimum_sale)

print("Top Product :", top_product)

print("Top Category :", top_category)

print("Top City :", top_city)

print("Total Orders :", total_orders)
```

Generate Business Insights

```python
insights = [
    f"1. Total business revenue generated is ${total_sales:,}.",

    f"2. The average sales per order is ${average_sales:.2f}.",

    f"3. '{top_product}' is the highest revenue-generating product.",

    f"4. '{top_city}' recorded the highest overall sales.",

    f"5. '{top_category}' is the best-performing product category."
]

print("===== BUSINESS INSIGHTS =====\n")

for insight in insights:
    print(insight)
```

Save Business Insight Report

```python
with open("business_insights.txt","w") as file:

    file.write("BUSINESS INSIGHT REPORT\n\n")

    for insight in insights:
        file.write(insight + "\n")

print("Insight report saved successfully.")
```

Export Business Summary

```python
summary = pd.DataFrame({
    "Metric":[
        "Total Sales",
        "Average Sales",
        "Highest Sale",
        "Lowest Sale",
        "Top Product",
        "Top Category",
        "Top City",
        "Total Orders"
    ],
    "Value":[
        total_sales,
        round(average_sales,2),
        maximum_sale,
        minimum_sale,
        top_product,
        top_category,
        top_city,
        total_orders
    ]
})

summary.to_csv("business_summary.csv", index=False)

print("Business summary exported.")
```
