Import Required Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

Create Sample Dataset

```python
import pandas as pd

data = {
    "Month": [
        "January",
        "February",
        "March",
        "April",
        "May",
        "June"
    ],
    "Sales": [
        120000,
        145000,
        160000,
        180000,
        175000,
        210000
    ],
    "Profit": [
        25000,
        30000,
        35000,
        42000,
        39000,
        50000
    ],
    "Category": [
        "Electronics",
        "Accessories",
        "Furniture",
        "Clothing",
        "Electronics",
        "Accessories"
    ]
}

df = pd.DataFrame(data)

df.to_csv("sales_data.csv", index=False)

print(df)
```

Load Dataset

```python
df = pd.read_csv("sales_data.csv")

df.head()
```

Display Dataset Information

```python
print(df.info())

print(df.describe())
```

Create Bar Chart

```python
plt.figure(figsize=(8,5))

plt.bar(df["Month"], df["Sales"])

plt.title("Monthly Sales")

plt.xlabel("Month")

plt.ylabel("Sales")

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig("bar_chart.png")

plt.show()
```

Create Line Chart

```python
plt.figure(figsize=(8,5))

plt.plot(df["Month"], df["Profit"], marker="o")

plt.title("Monthly Profit Trend")

plt.xlabel("Month")

plt.ylabel("Profit")

plt.grid(True)

plt.tight_layout()

plt.savefig("line_chart.png")

plt.show()
```

Create Pie Chart

```python
category_sales = df.groupby("Category")["Sales"].sum()

plt.figure(figsize=(7,7))

plt.pie(
    category_sales,
    labels=category_sales.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Sales Distribution by Category")

plt.tight_layout()

plt.savefig("pie_chart.png")

plt.show()
```
