Import Required Libraries

```python
import pandas as pd
import numpy as np
```

Create Sample Dataset

```python
import pandas as pd
import numpy as np

data = {
    "Employee_ID":[101,102,103,104,105,105,106,107,108,109],
    "Name":[
        "Rahul",
        "Priya",
        "Amit",
        np.nan,
        "Sneha",
        "Sneha",
        "Karan",
        "Pooja",
        "Riya",
        "Arjun"
    ],
    "Department":[
        "IT",
        "HR",
        "Finance",
        "IT",
        np.nan,
        np.nan,
        "Sales",
        "HR",
        "Finance",
        "Sales"
    ],
    "Age":[25,28,np.nan,30,26,26,29,np.nan,24,27],
    "Salary":[
        50000,
        45000,
        np.nan,
        60000,
        52000,
        52000,
        47000,
        46000,
        np.nan,
        51000
    ]
}

df = pd.DataFrame(data)

df.to_csv("raw_employee_data.csv", index=False)

print(df)
```

Load Dataset

```python
df = pd.read_csv("raw_employee_data.csv")

print(df.head())
```

Check Missing Values

```python
print("Missing Values")

print(df.isnull().sum())
```

Clean the Dataset

```python
# Fill missing values

df["Name"] = df["Name"].fillna("Unknown")

df["Department"] = df["Department"].fillna(df["Department"].mode()[0])

df["Age"] = df["Age"].fillna(df["Age"].mean())

df["Salary"] = df["Salary"].fillna(df["Salary"].median())

# Remove duplicate rows

df = df.drop_duplicates()

# Correct data types

df["Age"] = df["Age"].astype(int)

df["Salary"] = df["Salary"].astype(int)

print(df)
```

Verify Cleaned Dataset

```python
print(df.info())

print("\nRemaining Missing Values")

print(df.isnull().sum())
```

Export Cleaned Dataset

```python
df.to_csv("cleaned_employee_data.csv", index=False)

print("Cleaned dataset exported successfully.")
```

Download CSV (Google Colab)

```python
from google.colab import files

files.download("cleaned_employee_data.csv")
```
