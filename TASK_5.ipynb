Create Sample Dataset

````python
import pandas as pd
import numpy as np

data = {
    "Employee_ID":[101,102,103,104,105,105,106,107,108,109],
    "Name":[
        "Gopal",
        "Shweta",
        "Anu",
        "Sneha",
        np.nan,
        np.nan,
        "Karan",
        "Pooja",
        "Riya",
        "Arjun"
    ],
    "Age":[25,28,np.nan,30,26,26,29,np.nan,24,27],
    "Department":[
        "IT",
        "HR",
        "Finance",
        np.nan,
        "IT",
        "IT",
        "Sales",
        "HR",
        "Finance",
        "Sales"
    ],
    "Salary":[
        50000,
        45000,
        np.nan,
        60000,
        52000,
        52000,
        47000,
        46000,
        np.nan,
        51000
    ],
    "Joining_Date":[
        "2022-01-10",
        "2021-06-15",
        "2023-02-01",
        "2020-08-20",
        "2022-05-10",
        "2022-05-10",
        "2021-11-12",
        "2020-09-15",
        "2023-03-18",
        "2022-12-25"
    ]
}

df = pd.DataFrame(data)

df.to_csv("raw_employee_data.csv", index=False)

print(df)
````

Load Dataset

````python
import pandas as pd

df = pd.read_csv("raw_employee_data.csv")

df.head()
````

Check Dataset Information

````python
print(df.info())

print("\n")

print(df.describe())

print("\n")

print(df.isnull().sum())
````

Handle Missing Values

````python
df["Age"] = df["Age"].fillna(df["Age"].mean())

df["Salary"] = df["Salary"].fillna(df["Salary"].median())

df["Department"] = df["Department"].fillna(df["Department"].mode()[0])

df["Name"] = df["Name"].fillna("Unknown")

print(df.isnull().sum())
````

Remove Duplicate Records

````python
print("Rows Before:", len(df))

df = df.drop_duplicates()

print("Rows After:", len(df))
````

Correct Data Types

````python
print(df.dtypes)
````

Convert columns.

````python
df["Age"] = df["Age"].astype(int)

df["Salary"] = df["Salary"].astype(int)

df["Joining_Date"] = pd.to_datetime(df["Joining_Date"])

print(df.dtypes)
````

Final Clean Dataset

````python
print(df.head())
````

Save Clean Dataset

````python
df.to_csv("cleaned_employee_data.csv", index=False)

print("Clean dataset saved successfully.")
````
