Import Libraries

```python
# Import required libraries

import pandas as pd
import matplotlib.pyplot as plt
```

Create Sample Dataset

```python
# Create sample sales dataset

data = {
    "Month": [
        "January",
        "February",
        "March",
        "April",
        "May",
        "June"
    ],
    "Sales": [
        120000,
        145000,
        160000,
        180000,
        175000,
        210000
    ],
    "Profit": [
        25000,
        32000,
        38000,
        45000,
        42000,
        52000
    ],
    "Orders": [
        120,
        135,
        150,
        170,
        165,
        190
    ],
    "Category": [
        "Electronics",
        "Accessories",
        "Furniture",
        "Electronics",
        "Accessories",
        "Electronics"
    ]
}

df = pd.DataFrame(data)

df.to_csv("sales_data.csv", index=False)

print(df)
```

Load Dataset

```python
# Load dataset

df = pd.read_csv("sales_data.csv")

df.head()
```

Business Summary Function

```python
# Function to display business summary

def business_summary(data):

    print("="*40)
    print(" BUSINESS SUMMARY ")
    print("="*40)

    print("Total Sales :", data["Sales"].sum())
    print("Average Sales :", round(data["Sales"].mean(),2))
    print("Highest Sales :", data["Sales"].max())
    print("Lowest Sales :", data["Sales"].min())
    print("Total Orders :", data["Orders"].sum())
    print("Average Profit :", round(data["Profit"].mean(),2))

business_summary(df)
```

Professional Charts
Monthly Sales

```python
plt.figure(figsize=(8,5))

plt.bar(df["Month"], df["Sales"])

plt.title("Monthly Sales Analysis")

plt.xlabel("Month")

plt.ylabel("Sales")

plt.grid(axis="y")

plt.tight_layout()

plt.savefig("monthly_sales.png", dpi=300)

plt.show()
```

Profit Trend

```python
plt.figure(figsize=(8,5))

plt.plot(
    df["Month"],
    df["Profit"],
    marker="o",
    linewidth=2
)

plt.title("Monthly Profit Trend")

plt.xlabel("Month")

plt.ylabel("Profit")

plt.grid(True)

plt.tight_layout()

plt.savefig("profit_trend.png", dpi=300)

plt.show()
```

Category Distribution

```python
category = df.groupby("Category")["Sales"].sum()

plt.figure(figsize=(6,6))

plt.pie(
    category,
    labels=category.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Sales Distribution")

plt.tight_layout()

plt.savefig("category_distribution.png", dpi=300)

plt.show()
```

Business Insights

```python
print("BUSINESS INSIGHTS")

print("- Highest Sales Month :", df.loc[df["Sales"].idxmax(),"Month"])

print("- Lowest Sales Month :", df.loc[df["Sales"].idxmin(),"Month"])

print("- Total Revenue :", df["Sales"].sum())

print("- Average Profit :", round(df["Profit"].mean(),2))

print("- Total Orders :", df["Orders"].sum())
```

Export Clean Dataset

```python
df.to_csv("cleaned_sales_data.csv", index=False)

print("Dataset exported successfully.")
```
