import pandas as pd

````python
import pandas as pd

data = {
    "Month": ["January", "February", "March", "April", "May", "June"],
    "Sales": [120000, 145000, 160000, 180000, 175000, 210000],
    "Profit": [25000, 32000, 38000, 45000, 42000, 52000]
}

df = pd.DataFrame(data)

df.to_csv("sales_data.csv", index=False)

print(df)
````

Clean Dataset

````python
clean_df = df.copy()

clean_df.to_csv("cleaned_sales_data.csv", index=False)

print("Dataset cleaned successfully.")
````

Create Charts
Bar Chart

````python
plt.figure(figsize=(8,5))

plt.bar(clean_df["Month"], clean_df["Sales"])

plt.title("Monthly Sales")

plt.xlabel("Month")

plt.ylabel("Sales")

plt.tight_layout()

plt.savefig("bar_chart.png", dpi=300)

plt.show()
````

Line Chart

````python
plt.figure(figsize=(8,5))

plt.plot(clean_df["Month"], clean_df["Profit"], marker="o")

plt.title("Profit Trend")

plt.xlabel("Month")

plt.ylabel("Profit")

plt.grid(True)

plt.tight_layout()

plt.savefig("line_chart.png", dpi=300)

plt.show()
````

Pie Chart

````python
category = {
    "Electronics":55,
    "Furniture":25,
    "Accessories":20
}

plt.figure(figsize=(6,6))

plt.pie(
    category.values(),
    labels=category.keys(),
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Sales Distribution")

plt.savefig("pie_chart.png", dpi=300)

plt.show()
````

Generate Project Summary

````python
summary = f"""
PROJECT SUMMARY

Project Name:
GitHub Upload Practice

Total Records : {len(clean_df)}

Total Sales : {clean_df['Sales'].sum()}

Average Sales : {clean_df['Sales'].mean():,.2f}

Maximum Sales : {clean_df['Sales'].max()}

Minimum Sales : {clean_df['Sales'].min()}

Status:
Project Ready for GitHub Upload
"""

with open("project_summary.txt","w") as file:
    file.write(summary)

print(summary)
````
