Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

Create Sample Dataset

```python
import pandas as pd

data = {
    "Month": [
        "January",
        "February",
        "March",
        "April",
        "May",
        "June"
    ],
    "Sales": [
        120000,
        145000,
        160000,
        180000,
        175000,
        210000
    ],
    "Profit": [
        25000,
        32000,
        38000,
        45000,
        42000,
        52000
    ],
    "Orders": [
        120,
        135,
        150,
        170,
        165,
        190
    ],
    "Category": [
        "Electronics",
        "Accessories",
        "Furniture",
        "Electronics",
        "Accessories",
        "Electronics"
    ]
}

df = pd.DataFrame(data)

df.to_csv("sales_dashboard_data.csv", index=False)

print(df)
```

Load Dataset

```python
df = pd.read_csv("sales_dashboard_data.csv")

df.head()
```

Basic Business Analysis

```python
print("========== BUSINESS SUMMARY ==========\n")

print("Total Sales      :", df["Sales"].sum())

print("Average Sales    :", round(df["Sales"].mean(),2))

print("Maximum Sales    :", df["Sales"].max())

print("Minimum Sales    :", df["Sales"].min())

print("Total Orders     :", df["Orders"].sum())

print("Average Profit   :", round(df["Profit"].mean(),2))
```

Create Dashboard

```python
plt.figure(figsize=(16,10))

# -----------------------------
# Bar Chart
# -----------------------------
plt.subplot(2,2,1)

plt.bar(df["Month"], df["Sales"])

plt.title("Monthly Sales")

plt.xlabel("Month")

plt.ylabel("Sales")

plt.xticks(rotation=45)

# -----------------------------
# Line Chart
# -----------------------------
plt.subplot(2,2,2)

plt.plot(df["Month"], df["Profit"], marker="o")

plt.title("Monthly Profit")

plt.xlabel("Month")

plt.ylabel("Profit")

plt.grid(True)

# -----------------------------
# Pie Chart
# -----------------------------
plt.subplot(2,2,3)

category_sales = df.groupby("Category")["Sales"].sum()

plt.pie(
    category_sales,
    labels=category_sales.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Sales by Category")

# -----------------------------
# Business Summary
# -----------------------------
plt.subplot(2,2,4)

plt.axis("off")

summary = f"""

BUSINESS SUMMARY

Total Sales      : ₹{df['Sales'].sum():,}

Average Sales    : ₹{df['Sales'].mean():,.2f}

Maximum Sales    : ₹{df['Sales'].max():,}

Minimum Sales    : ₹{df['Sales'].min():,}

Total Orders     : {df['Orders'].sum()}

Average Profit   : ₹{df['Profit'].mean():,.2f}

Best Month       : {df.loc[df['Sales'].idxmax(),'Month']}

"""

plt.text(
    0,
    1,
    summary,
    fontsize=12,
    verticalalignment="top"
)

plt.suptitle(
    "Sales Performance Dashboard",
    fontsize=18,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig("sales_dashboard.png", dpi=300)

plt.show()
```
