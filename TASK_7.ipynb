Create Sample Business Dataset

```python
import pandas as pd

data = {
    "Order_ID": [1001,1002,1003,1004,1005,1006,1007,1008,1009,1010],
    "Product": [
        "Laptop","Mouse","Keyboard","Monitor","Laptop",
        "Headphone","Mouse","Laptop","Keyboard","Monitor"
    ],
    "Category": [
        "Electronics","Accessories","Accessories","Electronics",
        "Electronics","Accessories","Accessories","Electronics",
        "Accessories","Electronics"
    ],
    "Quantity": [2,5,3,1,4,6,2,1,5,2],
    "Price": [55000,700,1200,15000,56000,2500,750,54000,1300,15500],
    "Sales": [110000,3500,3600,15000,224000,15000,1500,54000,6500,31000],
    "City": [
        "Bhubaneswar","Cuttack","Puri","Rourkela","Bhubaneswar",
        "Cuttack","Balasore","Puri","Sambalpur","Bhubaneswar"
    ]
}

df = pd.DataFrame(data)

df.to_csv("sales_data.csv", index=False)

print(df)
```

Load Dataset

```python
df = pd.read_csv("sales_data.csv")

df.head()
```

Display Dataset Information

```python
print(df.info())

print("\nDataset Shape:", df.shape)
```

Calculate Total Values

```python
total_sales = df["Sales"].sum()
total_quantity = df["Quantity"].sum()

print("Total Sales :", total_sales)
print("Total Quantity Sold :", total_quantity)
```

Calculate Average Values

```python
average_sales = df["Sales"].mean()
average_price = df["Price"].mean()

print("Average Sales :", round(average_sales,2))
print("Average Price :", round(average_price,2))
```

Find Minimum Values

```python
print("Minimum Sales :", df["Sales"].min())
print("Minimum Price :", df["Price"].min())
```

Find Maximum Values

```python
print("Maximum Sales :", df["Sales"].max())
print("Maximum Price :", df["Price"].max())
```

Count Values

```python
print("Total Orders :", df["Order_ID"].count())

print("Product Count")

print(df["Product"].value_counts())
```

Group Data by Category

```python
category_summary = df.groupby("Category")["Sales"].sum()

print(category_summary)
```

City-wise Sales

```python
city_sales = df.groupby("City")["Sales"].sum()

print(city_sales)
```

Product-wise Sales

```python
product_sales = df.groupby("Product")["Sales"].sum()

print(product_sales)
```

Business Summary Table

```python
summary = pd.DataFrame({
    "Metric":[
        "Total Sales",
        "Average Sales",
        "Minimum Sale",
        "Maximum Sale",
        "Total Orders"
    ],
    "Value":[
        df["Sales"].sum(),
        round(df["Sales"].mean(),2),
        df["Sales"].min(),
        df["Sales"].max(),
        df["Order_ID"].count()
    ]
})

print(summary)
```

Save Business Summary

```python
summary.to_csv("business_summary.csv", index=False)

print("Business Summary Saved Successfully.")
```
